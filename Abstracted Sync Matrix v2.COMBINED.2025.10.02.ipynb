```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import Rectangle
import pytz
import random
from typing import Tuple

from datetime import datetime,timedelta,timezone

from astral import LocationInfo
from astral.sun import sun

import shapely.wkt as wkt
from shapely.geometry import Point

from timezonefinder import TimezoneFinder

################################################################################
################################################################################
## FUNCTIONS
################################################################################
################################################################################
##==============================================================================
## ADD HORIZONTAL GRID LINES
##==============================================================================

def add_horizontal_line(ax,y,config):
    """Add a horizontal line to a Matplotlib axes object using configuration parameters.

    Parameters
    ----------
    ax : matplotlib.axes.Axes
        The Matplotlib axes object to which the horizontal line will be added.
    y : float or int
        The y-coordinate where the horizontal line is drawn.
    config : dict
        Configuration dictionary containing line properties. Expected keys:
        - 'Line Width' : float or int
            Width of the line in points.
        - 'Line Style' : str
            Style of the line (e.g., '-' for solid, '--' for dashed, ':' for dotted).
        - 'Color' : str or tuple
            Color of the line (e.g., 'black', '#FF0000', (1.0, 0.0, 0.0)).
        - 'Alpha' : float
            Transparency of the line (0.0 for fully transparent, 1.0 for fully opaque).
        - 'zorder' : float or int
            The z-order of the line (controls stacking order).

    Returns
    -------
    matplotlib.lines.Line2D
        The Line2D object added to the axes.

    Notes
    -----
    This function uses `matplotlib.axes.Axes.axvline` to draw a horizontal line
    spanning the y-axis at the specified x-coordinate. Ensure the `config` dictionary
    contains all required keys ('Line Width', 'Line Style', 'Color', 'Alpha', 'zorder')
    with valid values to avoid KeyError or invalid rendering.

    Examples
    --------
    >>> import matplotlib.pyplot as plt
    >>> fig, ax = plt.subplots()
    >>> config = {"Line Width": 2,
                  "Line Style": "--",
                  "Color": "red",
                  "Alpha": 0.7,
                  "zorder": 5
                 }
    >>> line = add_vertical_line(ax, y = 5.0, config = config)
    >>> plt.show()
    """

    line_keys = ["Line Width","Line Style","Color","Alpha","zorder"]
    lw,ls,color,alpha,zo = [config[key] for key in line_keys]
    
    ax.axhline(y, 
               color = color, 
               alpha = alpha, 
               linewidth = lw, 
               linestyle = ls, 
               zorder = zo)
    
##==============================================================================
## ADD VERTICAL GRID LINES
##==============================================================================
    
def add_vertical_line(ax, x, config):
    """Add a vertical line to a Matplotlib axes object using configuration parameters.

    Parameters
    ----------
    ax : matplotlib.axes.Axes
        The Matplotlib axes object to which the vertical line will be added.
    x : float or int
        The x-coordinate where the vertical line is drawn.
    config : dict
        Configuration dictionary containing line properties. Expected keys:
        - 'Line Width' : float or int
            Width of the line in points.
        - 'Line Style' : str
            Style of the line (e.g., '-' for solid, '--' for dashed, ':' for dotted).
        - 'Color' : str or tuple
            Color of the line (e.g., 'black', '#FF0000', (1.0, 0.0, 0.0)).
        - 'Alpha' : float
            Transparency of the line (0.0 for fully transparent, 1.0 for fully opaque).
        - 'zorder' : float or int
            The z-order of the line (controls stacking order).

    Returns
    -------
    matplotlib.lines.Line2D
        The Line2D object added to the axes.

    Notes
    -----
    This function uses `matplotlib.axes.Axes.axvline` to draw a vertical line
    spanning the y-axis at the specified x-coordinate. Ensure the `config` dictionary
    contains all required keys ('Line Width', 'Line Style', 'Color', 'Alpha', 'zorder')
    with valid values to avoid KeyError or invalid rendering.

    Examples
    --------
    >>> import matplotlib.pyplot as plt
    >>> fig, ax = plt.subplots()
    >>> config = {"Line Width": 2,
                  "Line Style": "--",
                  "Color": "red",
                  "Alpha": 0.7,
                  "zorder": 5
                 }
    >>> line = add_vertical_line(ax, x = 0.5, config = config)
    >>> plt.show()
    """
    line_keys = ["Line Width", "Line Style", "Color", "Alpha", "zorder"]
    lw, ls, color, alpha, zo = [config[key] for key in line_keys]
   
    return ax.axvline(x,
                      color = color,
                      alpha = alpha,
                      linewidth = lw,
                      linestyle = ls,
                      zorder = zo)

##==============================================================================
## ADD BOX TO PLOT
##==============================================================================

def add_box(x,y,ls,lw,color,alpha,zo):
    
    plt.plot((x,x,x + 1,x + 1,x),(y,y + 1,y + 1,y,y),
             ls = "-",
             lw = 3.0,
             color = color,
             alpha = 0.99,
             zorder = 6.0)
    
    return

##==============================================================================
## ADD PATCH TO PLOT
##==============================================================================
    
def add_patch(ax, config):
    """
    Add patches to the plot based on given patch positions.

    Args:
        ax: Matplotlib axis object.
        config (tuples): (x_pos, y_pos, patch_width, patch_height, patch_color).
    """
    
    x,y = config["Lower Left Corner XY"]
        
    patch = Rectangle((x, y),
                      width = config["Banner Width"],
                      height = config["Banner Height"],
                      linewidth = config["Line Width"],
                      linestyle = config["Line Style"],
                      facecolor = config["Banner Face"],
                      edgecolor = config["Edge Color"],
                      alpha = config["Alpha"],
                      zorder = config["zorder"])
        
    ax.add_patch(patch)

##==============================================================================
## ADD ICOD
##==============================================================================

def add_text(ax, config):
    """Add text to a Matplotlib axes object using configuration parameters.

    Parameters
    ----------
    ax : matplotlib.axes.Axes
        The Matplotlib axes object to which the text will be added.
    config : dict
        Configuration dictionary containing text properties. Expected keys:
        - 'x' : float
            The x-coordinate of the text position.
        - 'y' : float
            The y-coordinate of the text position.
        - 'Text' : str
            The text string to display.
        - 'Horizontal Alignment' : str
            Horizontal alignment of the text ('left', 'center', 'right').
        - 'Vertical Alignment' : str
            Vertical alignment of the text ('top', 'center', 'bottom', 'baseline').
        - 'Font Size' : float or int
            Size of the text font.
        - 'Font Weight' : str or int
            Weight of the text font (e.g., 'normal', 'bold', 400, 700).
        - 'Font Color' : str or tuple
            Color of the text (e.g., 'black', '#FF0000', (1.0, 0.0, 0.0)).
        - 'Wrap' : bool
            Whether to wrap the text within the axes bounds.
        - 'zorder' : float or int
            The z-order of the text (controls stacking order).

    Returns
    -------
    matplotlib.text.Text
        The Text object added to the axes.

    Notes
    -----
    This function uses `matplotlib.axes.Axes.text` to place text on the plot.
    Ensure the `config` dictionary contains all required keys with valid values
    to avoid KeyError or invalid rendering.

    Examples
    --------
    >>> import matplotlib.pyplot as plt
    >>> fig, ax = plt.subplots()
    >>> config = {"x": 0.5,
                  "y": 0.5,
                  "Text": "Hello, World!",
                  "Horizontal Alignment": "center",
                  "Vertical Alignment": "center",
                  "Font Size": 12,
                  "Font Weight": "bold",
                  "Font Color": "blue",
                  "Wrap": True,
                  "zorder": 10
                 }
    >>> text_obj = add_text(ax, config)
    >>> plt.show()
    """
    
    return ax.text(x = config["x"],
                   y = config["y"],
                   s = config["Text"],
                   ha = config["Horizontal Alignment"],
                   va = config["Vertical Alignment"],
                   fontsize = config["Font Size"],
                   fontweight = config["Font Weight"],
                   color = config["Font Color"],
                   wrap = config["Wrap"],
                   zorder = config["zorder"])
    
##==============================================================================
## Add X & Y Axis Ticks & Labels
##==============================================================================

def add_axis_ticks(ax, axis_ticks: list, axis_labels: list, config: dict):
    """Set axis ticks and their properties on a Matplotlib axes object.

    Parameters
    ----------
    ax : matplotlib.axes.Axes
        The Matplotlib axes object to modify.
    axis_ticks : list
        List of tick positions (numeric values) for the specified axis.
    axis_labels : list
        List of tick labels (strings or numbers) corresponding to the tick positions.
    config : dict
        Configuration dictionary for axis and tick properties. Expected keys:
        - 'Axis' : str
            Axis to modify ('x' or 'y').
        - 'Which' : str
            Which ticks to modify ('major', 'minor', or 'both').
        - 'Direction' : str
            Tick direction ('in', 'out', or 'inout').
        - 'Length' : float or int
            Length of the tick marks in points.
        - 'Width' : float or int
            Width of the tick marks in points.
        - 'Color' : str or tuple
            Color of the tick marks (e.g., 'black', '#FF0000', (1.0, 0.0, 0.0)).
        - 'Left' : bool
            Whether to display ticks on the left side (for y-axis).
        - 'Right' : bool
            Whether to display ticks on the right side (for y-axis).
        - 'Top' : bool
            Whether to display ticks on the top side (for x-axis).
        - 'Bottom' : bool
            Whether to display ticks on the bottom side (for x-axis).
        - 'Label Size' : float or int
            Font size of the tick labels.
        - 'Label Color' : str or tuple
            Color of the tick labels (e.g., 'black', '#FF0000', (1.0, 0.0, 0.0)).
        - 'Label Rotation' : float or int
            Rotation angle of the tick labels in degrees.
        - 'Label Left' : bool
            Whether to display tick labels on the left side (for y-axis).
        - 'Label Right' : bool
            Whether to display tick labels on the right side (for y-axis).
        - 'Label Top' : bool
            Whether to display tick labels on the top side (for x-axis).
        - 'Label Bottom' : bool
            Whether to display tick labels on the bottom side (for x-axis).

    Returns
    -------
    None
        Modifies the axes object in place.

    Notes
    -----
    This function uses `matplotlib.axes.Axes.set_xticks` or `set_yticks` to set
    tick positions and labels, and `matplotlib.axes.Axes.tick_params` to configure
    tick and label properties. Ensure the `config` dictionary contains all required
    keys with valid values to avoid KeyError or invalid rendering. The `axis_ticks`
    and `axis_labels` lists must have the same length.

    Examples
    --------
    >>> import matplotlib.pyplot as plt
    >>> fig, ax = plt.subplots()
    >>> ticks = [0, 1, 2, 3]
    >>> labels = ['A', 'B', 'C', 'D']
    >>> config = {"Axis": "x",
                  "Which": "major",
                  "Direction": "out",
                  "Length": 6,
                  "Width": 1,
                  "Color": "black",
                  "Left": False,
                  "Right": False,
                  "Top": True,
                  "Bottom": True,
                  "Label Size": 10,
                  "Label Color": "blue",
                  "Label Rotation": 45,
                  "Label Left": False,
                  "Label Right": False,
                  "Label Top": False,
                  "Label Bottom": True
                 }
    >>> add_axis_ticks(ax, ticks, labels, config)
    >>> plt.show()
    """
    if config["Axis"] == "x":
        ax.set_xticks(axis_ticks, labels=axis_labels)
    else:
        ax.set_yticks(axis_ticks, labels=axis_labels)
       
    ax.tick_params(axis=config["Axis"],
                   which=config["Which"],
                   direction=config["Direction"],
                   length=config["Length"],
                   width=config["Width"],
                   color=config["Color"],
                   left=config["Left"],
                   right=config["Right"],
                   top=config["Top"],
                   bottom=config["Bottom"],
                   labelsize=config["Label Size"],
                   labelcolor=config["Label Color"],
                   labelrotation=config["Label Rotation"],
                   labelleft=config["Label Left"],
                   labelright=config["Label Right"],
                   labeltop=config["Label Top"],
                   labelbottom=config["Label Bottom"])
    
##==============================================================================
## ADD BANNER TO PLOT
##==============================================================================

def add_banner(ax, config):
        
    x,y = config["Lower Left Corner XY"]

    config["x"] = np.median([config["Banner Width"] + x, x])
    config["y"] = y + (config["Banner Height"]/2)
    
    print(config["Lower Left Corner XY"])
    
    add_patch(ax,config)
    add_text(ax, config)
    
##==============================================================================
## ADD ICOD
##==============================================================================

def add_icod(ax, config):
    """Add a UTC timestamp to a Matplotlib axes object using configuration parameters.

    Parameters
    ----------
    ax : matplotlib.axes.Axes
        The Matplotlib axes object to which the timestamp text will be added.
    config : dict
        Configuration dictionary containing text properties and date format. Expected keys:
        - 'Date Format' : str
            Format string for the timestamp (e.g., '%Y-%m-%d %H:%M:%S %Z').
        - Other keys as required by `add_text`:
            - 'x' : float
                The x-coordinate of the text position.
            - 'y' : float
                The y-coordinate of the text position.
            - 'Text' : str
                Placeholder for the timestamp text (overwritten by this function).
            - 'Horizontal Alignment' : str
                Horizontal alignment of the text ('left', 'center', 'right').
            - 'Vertical Alignment' : str
                Vertical alignment of the text ('top', 'center', 'bottom', 'baseline').
            - 'Font Size' : float or int
                Size of the text font.
            - 'Font Weight' : str or int
                Weight of the text font (e.g., 'normal', 'bold', 400, 700).
            - 'Font Color' : str or tuple
                Color of the text (e.g., 'black', '#FF0000', (1.0, 0.0, 0.0)).
            - 'Wrap' : bool
                Whether to wrap the text within the axes bounds.
            - 'zorder' : float or int
                The z-order of the text (controls stacking order).

    Returns
    -------
    matplotlib.text.Text
        The Text object added to the axes by `add_text`.

    Notes
    -----
    This function generates a UTC timestamp using `datetime.now(timezone.utc)` and formats
    it according to `config['Date Format']`. The timestamp is stored in `config['Text']` and
    passed to `add_text` to render on the axes. Ensure the `config` dictionary contains all
    required keys for `add_text` and a valid `'Date Format'` string to avoid errors.

    Examples
    --------
    >>> import matplotlib.pyplot as plt
    >>> from datetime import timezone
    >>> fig, ax = plt.subplots()
    >>> config = {"Date Format": "%Y-%m-%d @ %H%MZ",
                  "x": 0.5,
                  "y": 0.9,
                  "Text": "",  # Placeholder, overwritten by timestamp
                  "Horizontal Alignment": "center",
                  "Vertical Alignment": "top",
                  "Font Size": 10,
                  "Font Weight": "normal",
                  "Font Color": "black",
                  "Wrap": True,
                  "zorder": 10
                 }
    >>> text_obj = add_icod(ax, config)
    >>> plt.show()
    """
    
    config["Text"] = datetime.now(timezone.utc).strftime(config["Date Format"])
    
    return add_text(ax, config)

##==============================================================================
## ADD SHADED SUNRISE TO PLOT
##==============================================================================

def add_sun_shading(ax,
                    locality_point: Point,
                    sunrise_config: dict):
    """
    """

    print(str(locality_point))
    
    ## Calculate the Location information
    city = LocationInfo(locality_point.y,locality_point.x)
    
    ## SUNRISE/SUNSET: Calculate sunrise and sunset
    solar_data = sun(city.observer, date = now.today())

    print(solar_data)
    
    layers = sunrise_config["Number of Shading Layers"]
    shading_delta = sunrise_config["Shading Time Delta"]
    
    ## SUNRISE: Create a layered shading with a stepped time delta
    for sun_time in [solar_data["sunrise"] - timedelta(minutes = shading_delta * i) for i in range(layers)]:

        print("Sunrise:",sun_time)
        
        ## Convert datetime object to decimal time for plot
        sun_time_decimal = (sun_time.hour + sun_time.minute/60.0) * 2.0

        sunrise_config["Lower Left Corner XY"] = (0.0,-0.5)
        sunrise_config["Banner Width"] = sun_time_decimal
        sunrise_config["Banner Height"] = len(y_axis_labels)

        add_patch(ax,sunrise_config)

    ## SUNSET: Create a layered shading with a stepped time delta
    for sun_time in [solar_data["sunset"] + timedelta(minutes = shading_delta * i) for i in range(layers)]:

        print("Sunset:",sun_time)
        
        ## Convert datetime object to decimal time for plot
        sun_time_decimal = (sun_time.hour + sun_time.minute/60.0) * 2.0

        sunrise_config["Lower Left Corner XY"] = (sun_time_decimal,-0.5)
        sunrise_config["Banner Width"] = x_axis_max
        sunrise_config["Banner Height"] = len(y_axis_labels)

        add_patch(ax,sunrise_config)

##==============================================================================
## ADD END OF LOCAL DAY
##==============================================================================

def add_end_of_local_day(ax,
                         tz_offset: str,
                         config: dict):

    print(tz_offset)
    
    eod = 48 + (tz_offset * 2) if tz_offset < 0 else abs(tz_offset * 2)
    
    add_vertical_line(ax,eod,sync_config["End-of-Local Day Buffer"])
    add_vertical_line(ax,eod,sync_config["End-of-Local Day"])

def hm2hours_decimal(hours_min:str,
                     delimiter:str = ":") -> float:

    if delimiter in hours_min:
        hh,mm = [int(hm) for hm in hours_min.split(delimiter)]
        offset_decimal = hh + (mm / 60.0)

    else:
        print("Delimiter (%s) not found in %s." %(delimiter,hours_min))
        offset_decimal = 0.0

    return offset_decimal

def get_timezone_offset(point: Point, 
                        date: datetime = None) -> Tuple[str, str]:
    """Get timezone name and UTC offset for a Shapely Point geometry.

    Args:
        point: Shapely Point object with (x, y) as (longitude, latitude).
        date: Datetime object to check offset (default: current time).

    Returns:
        tuple[str, str]: (timezone_name, offset_str) e.g., ('America/Chicago', '-05:00').

    Raises:
        ValueError: If no timezone is found for the coordinates or offset cannot be determined.
    """
    
    tf = TimezoneFinder()
    lon, lat = point.x, point.y
    timezone_name = tf.timezone_at(lat=lat, lng=lon)
    
    if timezone_name is None:
        raise ValueError(f"No timezone found for coordinates ({lat}, {lon})")

    timezone = pytz.timezone(timezone_name)
    offset = timezone.utcoffset(date)

    print(offset)
    if offset is None:
        raise ValueError(f"Cannot determine offset for {timezone_name} at {date}")

    total_seconds = int(offset.total_seconds())
    hours, remainder = divmod(abs(total_seconds), 3600)
    minutes = remainder // 60
    offset_str = f"{'+' if total_seconds >= 0 else '-'}{hours:02d}:{minutes:02d}"

    offset_decimal = hm2hours_decimal(offset_str)
    
    return timezone_name, offset_str,offset_decimal
    
##==============================================================================
## UNPACK()
##==============================================================================

def unpack(packed_dictionary: dict) -> dict:
    """
    Convert a dictionary with lists of values as keys into a flattened dictionary.

    Args:
        packed_dictionary (dict): 
            A dictionary where each key maps to a list of values. For example:
            {
                "red": [0, 1, 2],
                "blue": [3, 4]
            }

    Returns:
        dict: 
            A dictionary where each value from the original lists becomes a key, 
            and its corresponding value is the key from the original dictionary.
            Example:
            {
                0: "red",
                1: "red",
                2: "red",
                3: "blue",
                4: "blue"
            }

    Example:
        >>> packed = {"red": [0, 1, 2], "blue": [3, 4]}
        >>> unpacked = unpack(packed)
        >>> print(unpacked)
        {0: "red", 1: "red", 2: "red", 3: "blue", 4: "blue"}
    """
    
    if isinstance(packed_dictionary, dict):
        unpacked_dictionary = dict()
        for key,val in packed_dictionary.items():
            for v in val:
                unpacked_dictionary[v] = key
            
        return unpacked_dictionary
    
    else:
        
        return packed_dictionary

plot_x = lambda t: ((t.hour) + (t.minute / 60.0) + (t.second / 3600.0)) * 2
    
################################################################################
################################################################################
## WORKFLOW DATA
################################################################################
################################################################################

iso8601 = "%Y-%m-%dT%H:%M:%SZ"
now = datetime.now(timezone.utc)

################################################################################
## DATA
################################################################################


################################################################################
## SYNC MATRIX CONFIGURATION
################################################################################

config_filename = "Config_PARTNER_Sync_Matrix.json"
config_filename = "Config_Abstracted_Sync_Matrix.json"
config_filename = "Config_NEW_Sync_Matrix.json"

with open(config_filename,"r") as f:
    sync_config = json.loads(f.read())

##==============================================================================
## SYNC MATRIX PARAMETERS & CONFIGURATION
## - Retreive Plot Configuration details from the Sync Config dictionary
##  - Colors
##  - ICOD Config
##  - Sunrise Config
##  - X-Axis Config
##  - Y-Axis Config
##==============================================================================

color_map = unpack(sync_config["Colors"])
icod_config = sync_config["ICOD Config"]
sunrise_config = sync_config["Sunrise/Sunset Config"]

##==============================================================================
## X-Axis Configuration & Parameters
## - X-Axis Config
## - X-Axis Min
## - X-Axis Max
## - X-Axis Labels
##==============================================================================

x_axis_config = sync_config["X-Axis Config"]

x_axis_min = sync_config["X-Axis Min"]
x_axis_max = sync_config["X-Axis Max"]
x_axis_ticks = np.arange(0,x_axis_max,1)
x_axis_labels = sync_config["X-Axis Labels"]

##==============================================================================
## Y-Axis Configuration & Parameters
##==============================================================================

y_axis_data = sync_config["Y-Axis Data"]

##------------------------------------------------------------------------------
## Calculate the Y-Axis minimum based on the Banner configurations in the 
## sync matrix
##------------------------------------------------------------------------------

if sync_config["Add Banners?"]:
    y_axis_min = np.min([val["Lower Left Corner XY"][1] for key,val in sync_config["Banners"].items()]) 
else:
    y_axis_min = sync_config["Y-Axis Min"]

##------------------------------------------------------------------------------
## Calculate Y-Axis Divisions based on the data grouping
##------------------------------------------------------------------------------

y_axis_divisions = [y_axis_min]
y_axis_labels = list()
increment = -0.5
for key,val in y_axis_data.items():
    ## Append the lenght of the Y-Axis sub-group to the
    ## Y-Axis divisions list
    increment += len(val)
    y_axis_divisions.append(increment)
    
    y_axis_labels += val

y_axis_config = sync_config["Y-Axis Config"]

y_axis_ticks = np.arange(0,len(y_axis_labels),1)
y_axis_max = len(y_axis_labels)

print(y_axis_divisions)
print(y_axis_labels)

##==============================================================================
## PLOT WIDTH & PLOT HEIGHT
##==============================================================================

## Retrieve User-defined Plot Width and Height from the Sync config dictionary
plot_width = sync_config["Plot Width"]
plot_height = len(y_axis_labels) * sync_config["Y-Axis Offset"]

################################################################################
################################################################################
## BODY
################################################################################
################################################################################
##==============================================================================
## Create a matplotlib plot object
##==============================================================================

## Define plot X-axis and Y-axis dimensions in inches
fig = plt.figure(figsize=(plot_width,plot_height))
ax = fig.add_subplot(111)

##==============================================================================
## XY Axis: Set Limits of the plot
##==============================================================================

ax.set_xlim(0, x_axis_max)
# ax.set_ylim(y_axis_min, y_axis_max - 0.5)

##==============================================================================
## XY Axis: Set Ticks & Labels
##==============================================================================

# ax.grid(True)
add_axis_ticks(ax,x_axis_ticks,x_axis_labels,x_axis_config)
add_axis_ticks(ax,y_axis_ticks,y_axis_labels,y_axis_config)

##==============================================================================
## Add Title to plot
##==============================================================================

plot_title = sync_config["Sync Matrix Title Stub"] + "\n%s" % now.strftime("%d %B %Y")
plt.title(plot_title,
          fontweight = "bold", 
          fontsize = 16, 
          color = "black")

##==============================================================================
## Add Vertical & Horizontal Grid Lines
##==============================================================================

for x in np.arange(0,x_axis_max,1.0):
    add_vertical_line(ax,x,sync_config["Grid Lines"])

for y in np.arange(-0.5,len(y_axis_labels),1):
    add_horizontal_line(ax,y,sync_config["Grid Lines"])

##==============================================================================
## Add Vertical & Horizontal Sub-Divider Lines for Y-Axis Categoricals
##==============================================================================

for x in [0,x_axis_max]:
    add_vertical_line(ax,x,sync_config["X-Axis Grouping"])

for y in y_axis_divisions:
    add_horizontal_line(ax,y,sync_config["Y-Axis Grouping"])

##==============================================================================
## Patches: Adding Data to Plot
##==============================================================================

# for position in patch_positions:
#     add_patch(ax,position)

##==============================================================================
## Add Banners to Plot
##==============================================================================

if sync_config["Add Banners?"]:
    for key,val in sync_config["Banners"].items():
        add_banner(ax, val)

    ## Add Box to Legend
    add_box(1.0,-1.75,"-",3.0,"orange",0.99,5)

##==============================================================================
## ADD DATA TO PLOT: Add Boxes to Plot
##==============================================================================
    
for i in range(0,20):

    x = np.random.randint(0,x_axis_max + 1)
    y = np.random.randint(0,y_axis_max)
    
    ax.scatter(x = x,
               y = y,
               marker = "o",
               s = 80,
               lw = 1.0,
               ls = "-",
               facecolor = "orange",
               edgecolor = "#525252",
               alpha = 0.8,
               zorder = 2)

for x,y,color_key in [(10,-1.0,"High"),(20,-1,"Moderate"),(30,-1,"Low"),(40,-1,"No Concern")]:

        ax.scatter(x = x,
               y = y,
               marker = "o",
               s = 80,
               lw = 1.0,
               ls = "-",
               facecolor = color_map.get(color_key,"pink"),
               edgecolor = "#525252",
               alpha = 0.8,
               zorder = 2)


date0 = now - timedelta(hours = 12)
eod0 = datetime(date0.year,date0.month,date0.day,23,59,59)
date1 = now + timedelta(hours = 12)

print("Start:", date0)
print("End:", date1)
print("EOD:", eod)

patch_config = {"Lower Left Corner XY": (0,y_axis_min), 
                "Banner Width": eod,
                "Banner Height": y_axis_max,
                "Banner Face": "#d9d9d9",
                "Edge Color": "#252525", 
                "Alpha": 0.5,
                "Line Width": 0.0,
                "Line Style": "-",
                "zorder": 4}

add_patch(ax,patch_config)

patch_config["Lower Left Corner XY"] = (plot_x(date1_patch_start) * 2,y_axis_min)
# add_patch(ax,patch_config) 
    
##==============================================================================
## SUNRISE/SUNSET: Define your location
##==============================================================================

tehran = wkt.loads("POINT(%f %f)" % (51.389722,35.688889))
london = wkt.loads("POINT(%f %f)" % (-0.1278,51.5074))
madrid = wkt.loads("POINT(%f %f)" % (-3.703333,40.416944))
nyc = wkt.loads("POINT(%f %f)" % (-74.006111,40.712778))
kathmandu = wkt.loads("POINT(%f %f)" % (85.32,27.71))
caracas = wkt.loads("POINT(%f %f)" % (-66.903611,10.480556))

if sync_config["Show Sunrise/Sunset?"]:
       
    ## Add sunrise and Sunset shading to Sync Matrix
    add_sun_shading(ax, london, sunrise_config)

##==============================================================================
## Add End-of-Local Day Vertical Line
## - Defined by Sync Config "Timezone Offset"
##==============================================================================

if sync_config["Show End of Local Day?"]:
    add_end_of_local_day(ax,tz_offset,sync_config)

##==============================================================================
## Add ICOD?
##==============================================================================

if sync_config["Show ICOD?"]:
    add_icod(ax, icod_config)

##==============================================================================
## Save Sync Matrix as a file
##==============================================================================

sync_matrix_filename = sync_config.get("Sync Matrix Filename","Default Sync Matrix.png")
if isinstance(sync_matrix_filename,list):
    sync_matrix_filename = sync_matrix_filename[0]
    
plt.savefig(sync_matrix_filename, dpi = 600, transparent = False, bbox_inches = "tight")

##==============================================================================
## Show Sync Matrix
##==============================================================================

plt.show()

plt.close()

################################################################################
################################################################################
## OUTPUTS
################################################################################
################################################################################
```
